```python
import os
import glob
import time
import random
import json
from concurrent.futures import ThreadPoolExecutor, as_completed
import numpy as np
from tqdm import tqdm
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.amp import autocast, GradScaler
from sklearn.model_selection import train_test_split
import torchvision.models as models
import albumentations as A
from albumentations.pytorch import ToTensorV2
import matplotlib.pyplot as plt

# =====================================================================
# 1. SETUP & CONFIGURATION
# =====================================================================

# Set your data path here
# DATA_PATH = "F:\\Facultate\\Licenta\\ARRAY_FORMAT"  # Change this to your Colab drive path when running on Colab
DATA_PATH = '/content/dataset_local'
class Config:
    DRIVE_PATH = '/content/drive'
    DATA_DIR = DATA_PATH  # Use the variable above
    CHECKPOINT_DIR = f'{DRIVE_PATH}/My Drive/Licenta/checkpoints'
    LOGS_DIR = f'{DRIVE_PATH}/My Drive/Licenta/logs'

    IN_CHANNELS = 1
    BATCH_SIZE = 8       # Reduced slightly for stability
    NUM_EPOCHS = 150
    LEARNING_RATE = 3e-4 # Lowered from 1e-3
    WEIGHT_DECAY = 1e-5
    MAX_GRAD_NORM = 1.0
    PATIENCE_EARLY_STOPPING = 150

    SEED = 42
    DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
    NUM_WORKERS = 2
    PIN_MEMORY = True

def seed_everything(seed=Config.SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

def setup_env():
    try:
        from google.colab import drive
        if not os.path.exists(Config.DRIVE_PATH):
            drive.mount(Config.DRIVE_PATH)
    except ImportError:
        pass
    os.makedirs(Config.CHECKPOINT_DIR, exist_ok=True)
    os.makedirs(Config.LOGS_DIR, exist_ok=True)
```

```python
# =====================================================================
# 3. MODEL DEFINITION
# =====================================================================

class DecoderBlock(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels):
        super().__init__()
        self.up = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels + skip_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels), nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels), nn.ReLU(inplace=True)
        )
    def forward(self, x, skip=None):
        x = self.up(x)
        if skip is not None:
            if x.shape[2:] != skip.shape[2:]:
                x = F.interpolate(x, size=skip.shape[2:], mode='bilinear', align_corners=True)
            x = torch.cat([x, skip], dim=1)
        return self.conv(x)

class MultiOutputUNet(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        resnet = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)
        weights = resnet.conv1.weight.clone()
        resnet.conv1 = nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3, bias=False)
        resnet.conv1.weight.data = torch.sum(weights, dim=1, keepdim=True)
        self.enc1 = nn.Sequential(resnet.conv1, resnet.bn1, resnet.relu)
        self.pool = resnet.maxpool
        self.enc2 = resnet.layer1
        self.enc3 = resnet.layer2
        self.enc4 = resnet.layer3
        self.enc5 = resnet.layer4
        self.dec4 = DecoderBlock(2048, 1024, 512)
        self.dec3 = DecoderBlock(512, 512, 256)
        self.dec2 = DecoderBlock(256, 256, 128)
        self.dec1 = DecoderBlock(128, 64, 64)
        self.final_conv = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True),
            nn.Conv2d(64, 32, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, num_classes, kernel_size=1)
        )
    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(e2)
        e4 = self.enc4(e3)
        e5 = self.enc5(e4)
        d4 = self.dec4(e5, e4)
        d3 = self.dec3(d4, e3)
        d2 = self.dec2(d3, e2)
        d1 = self.dec1(d2, e1)
        return self.final_conv(d1)
```

```python
# =====================================================================
# 4. LOSS & METRICS
# =====================================================================

class MultiOrganLoss(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        # per-organ pos_weight — ['artery', 'liver', 'stomach', 'vein'] (alfabetic)
        # artera e foarte mica, ii dam weight mult mai mare
        pos_weights = torch.tensor([150.0, 15.0, 20.0, 30.0]).view(1, -1, 1, 1).to(Config.DEVICE)
        self.bce = nn.BCEWithLogitsLoss(pos_weight=pos_weights)
        self.alpha = 0.4
        self.beta = 0.6
        self.gamma = 0.75

    def forward(self, logits, targets):
        bce_loss = self.bce(logits, targets)
        probs = torch.sigmoid(logits)
        smooth = 1e-5
        tp = (probs * targets).sum(dim=(2, 3))
        fp = ((1 - targets) * probs).sum(dim=(2, 3))
        fn = (targets * (1 - probs)).sum(dim=(2, 3))
        tversky = (tp + smooth) / (tp + self.alpha * fn + self.beta * fp + smooth)
        focal_tversky = torch.pow((1 - tversky), self.gamma).mean()
        return 0.4 * bce_loss + 0.6 * focal_tversky

def compute_metrics(logits, targets, organ_names):
    preds = (torch.sigmoid(logits) > 0.5).float()
    metrics = {organ: {} for organ in organ_names}

    TP = (preds * targets).sum(dim=(0, 2, 3))
    FP = (preds * (1 - targets)).sum(dim=(0, 2, 3))
    FN = ((1 - preds) * targets).sum(dim=(0, 2, 3))
    TN = ((1 - preds) * (1 - targets)).sum(dim=(0, 2, 3))

    smooth = 1e-7

    dice = (2 * TP) / (2 * TP + FP + FN + smooth)
    iou = TP / (TP + FP + FN + smooth)
    sensitivity = TP / (TP + FN + smooth)
    specificity = TN / (TN + FP + smooth)
    precision = TP / (TP + FP + smooth)

    for i, organ in enumerate(organ_names):
        metrics[organ]['dice'] = dice[i].item()
        metrics[organ]['iou'] = iou[i].item()
        metrics[organ]['sens'] = sensitivity[i].item()
        metrics[organ]['spec'] = specificity[i].item()
        metrics[organ]['prec'] = precision[i].item()

    return metrics
```

```python
# =====================================================================
# 2. DATASET
# =====================================================================

import re

def analyze_and_split_dataset(data_dir, val_ratio=0.15, test_ratio=0.15, seed=Config.SEED):
    all_files = sorted(glob.glob(os.path.join(data_dir, '*.npy')))
    if not all_files:
        raise FileNotFoundError(f"No .npy files found in {data_dir}")
    sample = np.load(all_files[0], allow_pickle=True).item()
    organ_names = sorted(sample['structures'].keys())

    patients = sorted(set(re.search(r'(P\d+)_IMG', os.path.basename(f)).group(1) for f in all_files))
    train_p, temp_p = train_test_split(patients, test_size=val_ratio + test_ratio, random_state=seed)
    val_p, test_p = train_test_split(temp_p, test_size=test_ratio / (val_ratio + test_ratio), random_state=seed)

    def get_split(p_set):
        return [f for f in all_files if re.search(r'(P\d+)_IMG', os.path.basename(f)).group(1) in p_set]

    train_paths = get_split(set(train_p))
    val_paths   = get_split(set(val_p))
    test_paths  = get_split(set(test_p))

    print(f"Organs: {organ_names}")
    print(f"Train: {len(train_paths)} | Val: {len(val_paths)} | Test: {len(test_paths)}")
    return train_paths, val_paths, test_paths, organ_names


class MultiOrganDataset(Dataset):
    def __init__(self, file_paths, organ_names, transform=None):
        self.paths = file_paths
        self.organ_names = organ_names
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        data = np.load(self.paths[idx], allow_pickle=True).item()

        # Eliminam slicing-ul [..., 0:1] care distrugea imaginea 2D
        image = np.ascontiguousarray(data['image'].astype(np.float32))
        masks_list = [np.ascontiguousarray(data['structures'][organ].astype(np.float32)) for organ in self.organ_names]

        if self.transform:
            augmented = self.transform(image=image, masks=masks_list)
            image = augmented['image']
            masks_list = augmented['masks']

            masks = torch.stack(
                [m if isinstance(m, torch.Tensor) else torch.from_numpy(np.ascontiguousarray(m)) for m in masks_list]
            ).float()
        else:
            masks = torch.stack([torch.from_numpy(m) for m in masks_list]).float()

        # Ne asiguram ca imaginea ajunge in format corect de Tensor
        if not isinstance(image, torch.Tensor):
            image = torch.from_numpy(np.ascontiguousarray(image))

        # Daca imaginea e inca 2D (H, W), ii adaugam canalul (1, H, W)
        if image.ndim == 2:
            image = image.unsqueeze(0)
        # Daca cumva e (H, W, 1), o facem (1, H, W)
        elif image.ndim == 3 and image.shape[-1] == 1:
            image = image.permute(2, 0, 1)

        return {'image': image.float(), 'masks': masks}
```

```python
# =====================================================================
# 5. TRANSFORMS  &  6. TRAINING
# =====================================================================

RESUME_TRAINING = False

setup_env()
seed_everything()

# 1. Load Data
if not os.path.exists(Config.DATA_DIR):
    fallback_dir = "/content/dataset_local"
    if os.path.exists(fallback_dir):
        print(f"WARNING: Config.DATA_DIR not found. Falling back to {fallback_dir}")
        Config.DATA_DIR = fallback_dir
    else:
        raise FileNotFoundError(f"Dataset directory not found: {Config.DATA_DIR}")

train_paths, val_paths, test_paths, organ_names = analyze_and_split_dataset(Config.DATA_DIR)
num_classes = len(organ_names)

import albumentations as A
from albumentations.pytorch import ToTensorV2

def get_transforms():
    train_transform = A.Compose([
        A.Resize(256, 256),
        A.HorizontalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.GaussNoise(p=0.2),
        A.ElasticTransform(alpha=1.0, sigma=50.0, p=0.2),
        # Aici este modificarea cheie:
        A.Normalize(mean=(0.5,), std=(0.5,), max_pixel_value=1.0),
        ToTensorV2()
    ], is_check_shapes=False)

    val_transform = A.Compose([
        A.Resize(256, 256),
        # Si aici:
        A.Normalize(mean=(0.5,), std=(0.5,), max_pixel_value=1.0),
        ToTensorV2()
    ], is_check_shapes=False)

    return train_transform, val_transform

train_transform, val_transform = get_transforms()

# 2. Datasets & Dataloaders
train_ds = MultiOrganDataset(train_paths, organ_names, train_transform)
val_ds   = MultiOrganDataset(val_paths,   organ_names, val_transform)

train_loader = DataLoader(train_ds, batch_size=Config.BATCH_SIZE, shuffle=True,
                          num_workers=Config.NUM_WORKERS, pin_memory=Config.PIN_MEMORY)
val_loader   = DataLoader(val_ds,   batch_size=Config.BATCH_SIZE, shuffle=False,
                          num_workers=Config.NUM_WORKERS, pin_memory=Config.PIN_MEMORY)

# 3. Model, Criterion, Optimizer, Scheduler, Scaler
model     = MultiOutputUNet(num_classes).to(Config.DEVICE)
criterion = MultiOrganLoss(num_classes)
optimizer = torch.optim.Adam(model.parameters(), lr=Config.LEARNING_RATE,
                             weight_decay=Config.WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=Config.NUM_EPOCHS)

scaler = GradScaler('cuda')

# Defaults
start_epoch = 1
best_val_dice = 0.0
patience_counter = 0
checkpoint_path = f"{Config.CHECKPOINT_DIR}/last_checkpoint.pth"

# 4. Resume Logic
if RESUME_TRAINING and os.path.exists(checkpoint_path):
    print(f"\n🔄 Resuming training from: {checkpoint_path}")
    try:
        checkpoint = torch.load(checkpoint_path, map_location=Config.DEVICE, weights_only=False)
    except Exception:
        print("⚠️ Standard torch.load failed, retrying with safe globals for NumPy scalars...")
        try:
            with torch.serialization.safe_globals([np._core.multiarray.scalar]):
                checkpoint = torch.load(checkpoint_path, map_location=Config.DEVICE, weights_only=False)
        except Exception as e:
            raise RuntimeError(
                f"Failed to load checkpoint {checkpoint_path}. "
                "Set RESUME_TRAINING = False if the file is corrupted or from an incompatible format."
            ) from e

    model.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
    scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
    start_epoch = checkpoint['epoch'] + 1
    best_val_dice = checkpoint['best_val_dice']
    patience_counter = checkpoint.get('patience_counter', 0)
    print(f"✅ State restored! Resuming at Epoch {start_epoch} (Best Dice: {best_val_dice:.4f})")
else:
    print(f"\n🚀 Starting training from scratch (Epoch 1)")


```

```python
# =====================================================================
# Vizualizare o singura imagine direct din DataLoader
# =====================================================================

# 1. Extragem primul batch din loader
# (Inlocuieste 'train_loader' cu numele variabilei tale, daca e diferit)
batch = next(iter(train_loader))

# 2. Luam prima imagine din batch (index 0).
# Daca loader-ul returneaza tuple in loc de dict, schimba in batch[0][0]
img_tensor = batch['image'][0]

# 3. O mutam pe CPU, scoatem dimensiunile in plus (1) si o facem array numpy
img_from_loader = img_tensor.cpu().squeeze().numpy()

# (Optional) Verificam forma si valorile pentru a fi siguri ca nu e "neagra"
print(f"Shape din loader: {img_from_loader.shape}")
print(f"Min: {img_from_loader.min():.4f}, Max: {img_from_loader.max():.4f}")

# 4. Plotarea in stilul tau (am pus doar 1 subplot in loc de 2)
fig, ax = plt.subplots(1, 1, figsize=(7, 6))
fig.patch.set_facecolor('#f0f0f0')
fig.suptitle("Test: Imagine direct din DataLoader", fontsize=12, fontweight='bold')

ax.imshow(img_from_loader, cmap='gray')
ax.set_title('Original din Batch', fontsize=13, fontweight='bold')
ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# =====================================================================
# LOOOOOOP
# =====================================================================


# 5. Training Loop
for epoch in range(start_epoch, Config.NUM_EPOCHS + 1):

    # --- TRAINING PHASE ---
    model.train()
    train_loss = 0.0
    pbar_train = tqdm(train_loader, desc=f"Epoch {epoch}/{Config.NUM_EPOCHS} [Train]", leave=False)

    for batch in pbar_train:
        images, masks = batch['image'].to(Config.DEVICE), batch['masks'].to(Config.DEVICE)

        optimizer.zero_grad()
        with autocast(Config.DEVICE):
            logits = model(images)
            loss = criterion(logits, masks)

        scaler.scale(loss).backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), Config.MAX_GRAD_NORM)
        scaler.step(optimizer)
        scaler.update()

        train_loss += loss.item()
        pbar_train.set_postfix_str(f"Loss: {loss.item():.4f}")

    train_loss /= len(train_loader)

    # --- VALIDATION PHASE ---
    model.eval()
    val_loss = 0.0
    val_stats = {organ: {'dice': [], 'iou': [], 'sens': [], 'spec': [], 'prec': []} for organ in organ_names}

    pbar_val = tqdm(val_loader, desc=f"Epoch {epoch}/{Config.NUM_EPOCHS} [Val  ]", leave=False)

    with torch.no_grad():
        for batch in pbar_val:
            images, masks = batch['image'].to(Config.DEVICE), batch['masks'].to(Config.DEVICE)

            with autocast(Config.DEVICE):
                logits = model(images)
                loss = criterion(logits, masks)

            val_loss += loss.item()
            metrics = compute_metrics(logits, masks, organ_names)

            for organ in organ_names:
                for k in val_stats[organ]:
                    val_stats[organ][k].append(metrics[organ][k])

    val_loss /= len(val_loader)

    # Calculate Averages
    avg_dice_all_organs = []
    print(f"\n{'-'*95}")
    print(f"EPOCH {epoch} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")
    print(f"{'-'*95}")
    print(f"{'Organ':<15} | {'Dice':<8} | {'IoU':<8} | {'Sens':<8} | {'Spec':<8} | {'Prec':<8}")
    print(f"{'-'*95}")

    for organ in organ_names:
        d = np.mean(val_stats[organ]['dice'])
        i = np.mean(val_stats[organ]['iou'])
        sn = np.mean(val_stats[organ]['sens'])
        sp = np.mean(val_stats[organ]['spec'])
        pr = np.mean(val_stats[organ]['prec'])

        avg_dice_all_organs.append(d)
        print(f"{organ:<15} | {d:.4f}   | {i:.4f}   | {sn:.4f}   | {sp:.4f}   | {pr:.4f}")

    avg_val_dice = np.mean(avg_dice_all_organs)
    print(f"{'-'*95}")
    print(f"MACRO AVG       | {avg_val_dice:.4f}")
    print(f"{'-'*95}\n")

    scheduler.step()

    # --- CHECKPOINTING ---
    torch.save({
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'best_val_dice': best_val_dice,
        'patience_counter': patience_counter
    }, checkpoint_path)

    if epoch in {30, 60, 90, 120, 150}:
        epoch_path = f"{Config.CHECKPOINT_DIR}/checkpoint_epoch_{epoch}.pth"
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'scheduler_state_dict': scheduler.state_dict(),
            'best_val_dice': best_val_dice,
            'patience_counter': patience_counter
        }, epoch_path)
        print(f"💾 Saved scheduled checkpoint: {epoch_path}")

    if avg_val_dice > best_val_dice:
        best_val_dice = avg_val_dice
        patience_counter = 0
        torch.save({'model_state_dict': model.state_dict()}, f"{Config.CHECKPOINT_DIR}/best_model.pt")
        print("🌟 Saved new best model!")
    else:
        patience_counter += 1

    if patience_counter >= Config.PATIENCE_EARLY_STOPPING:
        print("🛑 Early Stopping triggered!")
        break

print("\n✅ Training Complete.")
```

```python
# =====================================================================
# 6. EVALUATION
# =====================================================================

# Load the best model for evaluation

# best_model_path = f"{Config.CHECKPOINT_DIR}/best_model.pth"
# checkpoint = torch.load(best_model_path, map_location=Config.DEVICE, weights_only=False)
# model.load_state_dict(checkpoint['model_state_dict'])
# print(f"✅ Loaded epoch {checkpoint['epoch']}, best val dice: {checkpoint['best_val_dice']:.4f}")

best_model_path = f"{Config.CHECKPOINT_DIR}/best_model.pt"

if os.path.exists(best_model_path):
    checkpoint = torch.load(best_model_path, map_location=Config.DEVICE)
    if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
        checkpoint = checkpoint['model_state_dict']
    model.load_state_dict(checkpoint)
    print("Loaded best model for evaluation.")
else:
    print("Best model not found, using current model.")

# Create test dataloader
test_ds = MultiOrganDataset(test_paths, organ_names, val_transform)
test_loader = DataLoader(test_ds, batch_size=Config.BATCH_SIZE, shuffle=False,
                         num_workers=Config.NUM_WORKERS, pin_memory=Config.PIN_MEMORY)

# Evaluate on test set
model.eval()
test_loss = 0.0
test_stats = {organ: {'dice': [], 'iou': [], 'sens': [], 'spec': [], 'prec': []} for organ in organ_names}

with torch.no_grad():
    for batch in tqdm(test_loader, desc="Evaluating on Test Set"):
        images, masks = batch['image'].to(Config.DEVICE), batch['masks'].to(Config.DEVICE)

        with autocast(Config.DEVICE):
            logits = model(images)
            loss = criterion(logits, masks)

        test_loss += loss.item()
        metrics = compute_metrics(logits, masks, organ_names)

        for organ in organ_names:
            for k in test_stats[organ]:
                test_stats[organ][k].append(metrics[organ][k])

test_loss /= len(test_loader)

print(f"\n{'='*95}")
print("TEST RESULTS")
print(f"{'='*95}")
print(f"Test Loss: {test_loss:.4f}")
print(f"{'-'*95}")
print(f"{'Organ':<15} | {'Dice':<8} | {'IoU':<8} | {'Sens':<8} | {'Spec':<8} | {'Prec':<8}")
print(f"{'-'*95}")

avg_dice_test = []
for organ in organ_names:
    d = np.mean(test_stats[organ]['dice'])
    i = np.mean(test_stats[organ]['iou'])
    sn = np.mean(test_stats[organ]['sens'])
    sp = np.mean(test_stats[organ]['spec'])
    pr = np.mean(test_stats[organ]['prec'])

    avg_dice_test.append(d)
    print(f"{organ:<15} | {d:.4f}   | {i:.4f}   | {sn:.4f}   | {sp:.4f}   | {pr:.4f}")

avg_test_dice = np.mean(avg_dice_test)
print(f"{'-'*95}")
print(f"MACRO AVG       | {avg_test_dice:.4f}")
print(f"{'='*95}")
```

```python
# =====================================================================
# 7. VISUALIZATION
# =====================================================================

def plot_results(model, dataloader, organ_names, random_seed=None, max_organs=4):
    model.eval()
    sample = None

    if random_seed is not None:
        np.random.seed(random_seed)
        random.seed(random_seed)

    with torch.no_grad():
        for batch in dataloader:
            images = batch['image'].to(Config.DEVICE)
            masks = batch['masks'].to(Config.DEVICE)
            logits = model(images)
            probs = torch.sigmoid(logits)
            preds = (probs > 0.5).float()

            batch_size = images.shape[0]
            idx = np.random.randint(batch_size)
            sample = {
                'image': images[idx].cpu().squeeze(0).numpy(),
                'mask': masks[idx].cpu().numpy(),
                'pred': preds[idx].cpu().numpy()
            }
            break

    if sample is None:
        raise ValueError('No samples were found in the dataloader.')

    img = sample['image']
    gt_masks = sample['mask']
    pred_masks = sample['pred']
    num_organs = gt_masks.shape[0]
    organ_indices = list(range(min(num_organs, max_organs)))
    selected_organs = [organ_names[i] for i in organ_indices]

    cols = len(selected_organs)
    fig, axes = plt.subplots(3, cols, figsize=(4 * cols, 12))
    if cols == 1:
        axes = np.expand_dims(axes, axis=1)

    for col_idx, organ_idx in enumerate(organ_indices):
        organ_name = organ_names[organ_idx]
        gt_mask = gt_masks[organ_idx]
        pred_mask = pred_masks[organ_idx]

        axes[0, col_idx].imshow(img, cmap='gray')
        axes[0, col_idx].set_title(f'Test Image: {organ_name}')
        axes[0, col_idx].axis('off')

        axes[1, col_idx].imshow(gt_mask, cmap='Greens')
        axes[1, col_idx].set_title('Expected Mask (GT)')
        axes[1, col_idx].axis('off')

        axes[2, col_idx].imshow(pred_mask, cmap='Reds')
        axes[2, col_idx].set_title('Model Prediction')
        axes[2, col_idx].axis('off')

        intersection = np.logical_and(gt_mask > 0.5, pred_mask > 0.5).sum()
        union = gt_mask.sum() + pred_mask.sum()
        dice = (2.0 * intersection) / (union + 1e-7) if union > 0 else 1.0
        axes[2, col_idx].set_xlabel(f'Dice: {dice:.4f}', fontsize=10)

    plt.tight_layout()
    plt.show()

# Plot one random example from the test set
plot_results(model, test_loader, organ_names, random_seed=3, max_organs=4)
```
